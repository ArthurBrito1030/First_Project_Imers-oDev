```python
df.columns
```

```python
mapeamento_colunas = {
    'work_year': 'ano',
    'experience_level': 'senioridade',
    'employment_type': 'contrato',
    'job_title': 'cargo',
    'salary': 'salario',
    'salary_currency': 'moeda',
    'salary_in_usd': 'usd',
    'employee_residence': 'residencia',
    'remote_ratio': 'remoto',
    'company_location': 'empresa',
    'company_size': 'tamanho_empresa'
}

df = df.rename(columns=mapeamento_colunas)
df.columns
```

```python
df["senioridade"].value_counts()
```

```python
df["contrato"].value_counts()
```

```python
df["remoto"].value_counts()
```

```python
df["tamanho_empresa"].value_counts()
```

```python
senioridade = {
    'SE': 'Sênior',
    'MI': 'Pleno',
    'EN': 'Júnior ',
    'EX': 'Executivo'
}

df['senioridade'] = df['senioridade'].replace(senioridade)
df['senioridade'].value_counts()
```

```python
mapeamento_contrato = {
    'FT': 'Tempo Integral',
    'CT': 'Contrato',
    'PT': 'Meio Período',
    'FL': 'Freelancer'
}

df['contrato'] = df['contrato'].replace(mapeamento_contrato)

df['contrato'].value_counts()
```

```python
remoto = {
    0: 'Presencial',
    50: 'Híbrido',
    100: 'Remoto'
}

df['remoto'] = df['remoto'].replace(remoto)
df['remoto'].value_counts()
```

```python
tamanho_empresa = {
    'M': 'Médio',
    'L': 'Grande',
    'S': 'Pequeno'
}

df['tamanho_empresa'] = df['tamanho_empresa'].replace(tamanho_empresa)
df['tamanho_empresa'].value_counts()
```

```python
import numpy as np

# Criação de dataframe de teste para usar de exemplo
df_salarios = pd.DataFrame({
      'nome': ["Arthur", "Sarah", "Betina", "Henrique", "Caio"],
      'salarios': [8000, np.nan, 7000, np.nan, 40000]

})

# Calcula a média salarial e substitui os nulos pela mpedia e arredonda os valores
df_salarios['salarios_media'] = df_salarios['salarios'].fillna(df_salarios['salarios'].mean().round(2))
df_salarios['salarios_mediana'] = df_salarios['salarios'].fillna(df_salarios['salarios'].median())
df_salarios
```

```python
df_temperaturas = pd.DataFrame({
    "Dias": ["Segunda", "Terça", "Quarta", "Quinta", "Sexta"],
    "Temperatura": [27, np.nan, np.nan, 26, 25]
})

df_temperaturas["preenchido_ffill"] = df_temperaturas["Temperatura"].ffill()
df_temperaturas
```

```python
df_temperaturas = pd.DataFrame({
    "Dias": ["Segunda", "Terça", "Quarta", "Quinta", "Sexta"],
    "Temperatura": [27, np.nan, np.nan, 26, 25]
})

df_temperaturas["preenchido_bfill"] = df_temperaturas["Temperatura"].bfill()
df_temperaturas
```

```python
df_cidades = pd.DataFrame({
    'nome': ["Arthur", "Sarah", "Betina", "Henrique", "Caio"],
    'cidade': ["São Paulo", np.nan, "Campinas",np.nan,"Belém"],
})

df_cidades["cidades_preenchidas"] = df_cidades["cidade"].fillna("Não informado")
df_cidades
```

# Aula3-Imersão Python

```python
df_limpo.head()
```

```python
df_limpo['senioridade'].value_counts().plot(kind="bar", title="Distribuição de Senioridade")
```

```python
import seaborn as sns
```

```python
sns.barplot (data=df_limpo, x= "senioridade", y= "usd")
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(7, 5))
sns.barplot(data=df_limpo, x="senioridade", y="usd")
plt.title("Salário médio por senioridade")
plt.xlabel("senioridade")
plt.ylabel("Salário anual (USD)")
plt.show()
```

```python
df_limpo.groupby("senioridade")["usd"].mean().sort_values(ascending=False)
```

```python
ordem = df_limpo.groupby("senioridade")["usd"].mean().sort_values(ascending=False).index
ordem
```

```python
plt.figure(figsize=(7, 5))
sns.barplot(data=df_limpo, x="senioridade", y="usd", order=ordem)
plt.title("Salário médio por senioridade")
plt.xlabel("senioridade")
plt.ylabel("Salário anual (USD)")
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(df_limpo['usd'], bins = 50, kde=True)
plt.title("Distribuição dos Salários Anuais")
plt.xlabel("Salário anual (USD)")
plt.ylabel("Frequência")
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.boxplot(x=df_limpo['usd'])
plt.title("Boxplot Salários")
plt.xlabel("Salário anual (USD)")
plt.show()
```

```python
ordem_senioridade = ["Sênior", "Pleno", "Júnior ", "Executivo"]

plt.figure(figsize=(8, 5))
sns.boxplot(x='senioridade', y='usd', data=df_limpo, order=ordem_senioridade)
plt.title("Boxplot da Distribuição por Senioridade")
plt.xlabel("Senioridade")
plt.show()
```
